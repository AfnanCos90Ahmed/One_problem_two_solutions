```python
class ShoppingCart:

    # Part A
    def __init__(self, customer_name, items=None):
        self.customer_name = customer_name
        self.items = items if items is not None else []

    # Part B
    def add_item(self, price):
        if price > 0:
            self.items.append(price)
        else:
            print("Price must be positive")

    def remove_item(self, price):
        if price in self.items:
            self.items.remove(price)
        else:
            print("Item not found")

    def calculate_total(self):
        total = 0
        for price in self.items:
            total += price
        return total

    def apply_discount(self):
        total = self.calculate_total()

        if total >= 3000:
            return total * 0.90
        else:
            return total

    # Part C
    def display_cart(self):
        print("Customer Name:", self.customer_name)
        print("Items:", self.items)
        print("Total Amount:", self.calculate_total())
        print("Final Amount:", self.apply_discount())
        print("Number of Items:", len(self.items))

    def clear_cart(self):
        self.items.clear()
        print("Cart cleared successfully!")

    # Part D
    def __add__(self, other):
        new_name = self.customer_name + " & " + other.customer_name
        new_items = self.items + other.items

        return ShoppingCart(new_name, new_items)


# Part E

cart1 = ShoppingCart("Arisha", [100, 200, 300])
cart2 = ShoppingCart("Zarisha", [300, 400, 500])

cart1.add_item(600)
cart2.add_item(550)

cart1.remove_item(600)
cart2.remove_item(300)

print("---- CART 1 ----")
cart1.display_cart()

print("\n---- CART 2 ----")
cart2.display_cart()

cart3 = cart1 + cart2

print("\n---- COMBINED CART ----")
cart3.display_cart()
```
